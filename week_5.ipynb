```python
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

df = pd.read_csv("/content/drive/MyDrive/Indian_Kids_Screen_Time.csv")

df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
bins = [5,10,13,16,19]
labels = ['6-10','11-13','14-16','17-18']

df['Age_Band'] = pd.cut(df['Age'], bins=bins, labels=labels)

df.head()
```

```python
cohort = pd.crosstab(df['Age_Band'], df['Primary_Device'])

cohort
```

```python
cohort.plot(kind='bar', figsize=(10,6))

plt.title("Device Usage by Age Cohort")
plt.xlabel("Age Band")
plt.ylabel("Number of Users")
plt.legend(title="Device")
plt.show()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(cohort, annot=True, cmap='YlGnBu')

plt.title("Heatmap: Age Cohort vs Device Type")
plt.show()
```

```python
segment = pd.crosstab(df['Urban_or_Rural'], df['Primary_Device'])

segment
```

```python
segment.plot(kind='bar', stacked=True, figsize=(8,6))

plt.title("Device Usage by Location Segment")
plt.xlabel("Location")
plt.ylabel("Count")
plt.show()
```

```python
sns.boxplot(x='Gender', y='Avg_Daily_Screen_Time_hr', data=df)

plt.title("Screen Time Distribution by Gender")
plt.show()
```
