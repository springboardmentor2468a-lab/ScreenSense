```python
from google.colab import drive
import pandas as pd

drive.mount('/content/drive')

df = pd.read_csv("/content/drive/MyDrive/Indian_Kids_Screen_Time.csv")

df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="whitegrid")
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x="Age", y="Avg_Daily_Screen_Time_hr", data=df)

plt.title("Average Daily Screen Time by Age")
plt.xlabel("Age")
plt.ylabel("Screen Time (hours)")
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.boxplot(x="Gender", y="Avg_Daily_Screen_Time_hr", data=df)

plt.title("Screen Time Distribution by Gender")
plt.show()
```

```python
plt.figure(figsize=(6,5))

sns.barplot(x="Urban_or_Rural", y="Avg_Daily_Screen_Time_hr", data=df)

plt.title("Average Screen Time: Urban vs Rural")
plt.show()
```

```python
device_count = df["Primary_Device"].value_counts()

device_count.plot(kind="bar")

plt.title("Most Used Devices by Kids")
plt.xlabel("Device Type")
plt.ylabel("Number of Users")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x="Health_Impacts", y="Avg_Daily_Screen_Time_hr", data=df)

plt.title("Health Impact vs Screen Time")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.barplot(data=df, x="Age", y="Avg_Daily_Screen_Time_hr", estimator='mean')

plt.title("Average Screen Time by Age")
plt.xlabel("Age")
plt.ylabel("Average Daily Screen Time (hours)")

plt.show()
```
