```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("cleaned_Indian_Kids_Screen_Time.csv")
```

```python
df.columns
```

```python
df.head()
```

# Habit Patterns

### Weekday vs Weekend Screen TIme

```python
sns.boxplot(x='weekday_flag', y='avg_daily_screen_time_hr', data=df)
plt.title("Screen Time: Weekday vs Weekend")
plt.xlabel("Weekday or Weekend")
plt.ylabel("Average Daily Screen Time (Hours)")
plt.show()
```

### Insight
- Screen time is higher during weekends compared to weekdays.
- Users have more free time, leading to increased device usage for recreational.
### Possible Drive
- Users have more free time on weekends

### Activity Type by Weekday/Weekend

```python

sns.countplot(x='weekday_flag', hue='activity_type', data=df)
plt.title("Activity Type: Weekday vs Weekend")
plt.xlabel("Weekday or Weekend")
plt.ylabel("Number of Users")
plt.show()
```

### Insight
- Recreational usage slightly increases during weekends compared to weekdays.
- This indicates users prefer recreational activities during weekends
### Possible Drivers
- Weekdays are mainly focused on school or study activities.
- Weekends provide more free time for games, videos, and social media

### Device Usage by Weekday/Weekend

```python
sns.countplot(x='weekday_flag', hue='primary_device', data=df)
plt.title("Device Usage: Weekday vs Weekend")
plt.xlabel("Weekday or Weekend")
plt.ylabel("Number of Users")
plt.show()
```

### Insight
- Smartphones are the most commonly used device across all users.
- Device usage slightly increases during weekends.
- Laptop usage increases slightly during weekends compared to weekdays
### Possible Drivers
- Smartphones are portable and easily accessible, making them the most preferred device.
- Weekends provide more free time, leading to increased laptop usage for entertainment or learning.
- TV usage stays consistent due to regular viewing habits.

# Segment Wise _Insights

### Device Usage by Location

```python
sns.countplot(x='urban_or_rural', hue='primary_device', data=df)
plt.title("Device Usage by Location")
plt.xlabel("Location Type")
plt.ylabel("Number of Users")
plt.show()
```

### Insight
- Urban users show higher device usage than rural users.
- Smartphones dominate in both locations.
- Rural areas have lower usage of laptops and tablets.
### Possible Drivers:
- This may be due to better internet access and higher device availability in urban areas.
- Rural areas may have limited access to digital devices.


### Screen Time by Age Group

```python
sns.boxplot(x='age_group', y='avg_daily_screen_time_hr', data=df)
plt.title("Screen Time Distribution by Age Group")
plt.xlabel("Age Band")
plt.ylabel("Average Daily Screen Time (Hours)")
plt.show()
```

### Insight
- Young teens (13–15) have relatively higher median screen time compared to the other groups.
- Children (8–12) show greater variation and several high outliers, indicating that some children spend very long hours on screens.
-  Teens (16–18) generally show moderate screen usage, with fewer extreme values

### Activity Type by Gender

```python
sns.countplot(x='gender', hue='activity_type', data=df)
plt.title("Activity Type by Gender Segment")
plt.xlabel("Gender")
plt.ylabel("Number of Users")
plt.show()
```

### Insight
- Recreational screen activity is almost similar for both male and female users.
- Male users show a slightly higher count of recreational usage compared to females.
### Possible Drivers
- Both genders have similar access to digital devices and entertainment platforms.
- Slight differences may be due to personal interests in games, videos, or social media.

### Screen Time By Location 

```python
sns.boxplot(x='urban_or_rural', y='avg_daily_screen_time_hr', data=df)
plt.title("Screen Time by Location Segment")
plt.xlabel("Location Type")
plt.ylabel("Average Daily Screen Time (Hours)")
plt.show()
```

### Insight 
- Average screen time is slightly higher in urban areas compared to rural areas.
- Both locations show similar usage patterns, but urban users have more high screen-time values (outliers).
### Possible Drivers
- Better internet connectivity and device availability in urban areas.
- Rural areas may have limited access to digital devices and internet services.
