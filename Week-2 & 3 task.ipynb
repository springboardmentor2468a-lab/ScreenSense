```python
def donut_chart(column_name, title):
    counts = df[column_name].value_counts()

    plt.figure()

    wedges, texts, autotexts = plt.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        pctdistance=0.75,      # Move percentage inside
        labeldistance=1.1      # Keep labels slightly outside
    )

    # Create white circle for donut
    centre_circle = plt.Circle((0, 0), 0.60, fc='white')
    fig = plt.gcf()
    fig.gca().add_artist(centre_circle)

    plt.title(title)
    plt.show()
```
