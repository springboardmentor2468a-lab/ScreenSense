# Insights

#Unvariant Analysis
1. Age Distribution
Most children in the dataset fall within the middle age groups, meaning the data mainly represents school-age students. Younger and older age groups appear less frequently.

2. Screen Time Distribution
Most children spend a moderate amount of time on screens daily, while only a small number of children have extremely high screen usage.

3. Gender Distribution
The dataset contains a fairly balanced representation of male and female participants.

4. Primary Device Distribution
Portable devices such as mobile phones and tablets appear to be the most commonly used devices among children compared to stationary devices.

5. Educational to Recreational Ratio
The screen usage pattern suggests that children generally use screens more for recreational purposes than for educational activities.

6. Exceeded Recommended Limit
A noticeable portion of children exceed the recommended daily screen time limit, indicating potential concerns regarding excessive screen exposure.

7. Health Impacts Distribution
Common health issues related to screen usage include eye strain, headaches, and sleep disturbances.

8. Urban vs Rural Distribution
The dataset includes children from both urban and rural areas, allowing comparison between different living environments.

9. Age Band Distribution
Age bands categorize children into developmental stages, helping to analyze screen habits across different age groups.

10. Device Type Category
Portable devices account for the majority of screen usage, suggesting that handheld devices are more accessible and frequently used by children.

11. Usage Level Distribution
Most children fall into moderate or high screen usage categories, while fewer children belong to the low usage group.

# Bivariant Analysis
1. Age vs Screen Time
Screen time tends to increase slightly as age increases, suggesting that older children spend more time using digital devices.

2. Gender vs Screen Time
Comparison between genders shows potential differences in average screen time usage patterns.

3. Device vs Screen Time
Portable devices are often associated with higher screen time usage due to their convenience and accessibility.

4. Urban vs Screen Time
Children living in urban areas may have higher screen exposure compared to rural children because of greater access to technology and internet connectivity.

5. Device Type vs Exceeded Limit
Children using portable devices are more likely to exceed recommended screen time limits.

6. Age Band vs Usage Level
Higher age bands tend to show higher levels of screen usage, indicating increased screen dependency as children grow older.

7. Correlation Matrix
The correlation matrix highlights relationships between numerical variables such as age, screen time, and educational-to-recreational usage ratio.
