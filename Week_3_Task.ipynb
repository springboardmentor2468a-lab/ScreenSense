```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("cleaned_Indian_Kids_Screen_Time.csv")
```

```python
df.columns
```

# Univariate Analysis

### Distribution Of Daily Screen Time

```python
#Histogram
sns.histplot(df['avg_daily_screen_time_hr'], bins=10, kde=True)
plt.title("Distribution of Daily Screen Time")
plt.xlabel("Daily Screen Time (hours)")
plt.ylabel("Number Of User")
plt.show()
```

### Insight:
- Most users spend 5-6 hours daily on screen.
- Peak usage is around 5 hours.
- Very few users spend less than 2 hours.
- Some users spend 8+ hours, indicating possible heavy usage/outliers

### Age Band Distribution

```python
#Bar Chart
sns.countplot(x='age_group', data=df)
plt.xlabel("Age Band")
plt.ylabel("Number Of User")
plt.title("Distribution of Age Bands")
plt.show()
```

### Insight
The 8–12 children and 16–18 teens age groups have the highest number of users, while the **13–15 young teens group has slightly fewer users

### Device Usage

```python
#Bar Chart
sns.countplot(x='primary_device', data=df)
plt.xlabel("Primary Device")
plt.ylabel("Number Of User")
plt.title("Device Usage Distribution")
plt.show()
```

### Insight
Smartphones are the most commonly used device, followed by TV, while laptops and tablets have fewer users

# Bivariate Analysis

### Screen time by gender

```python
#boxplot
sns.boxplot(x='gender', y='avg_daily_screen_time_hr', data=df)
plt.xlabel('gender')
plt.ylabel("Avg daily screen time(Hours)")
plt.title("Screen Time by Gender")
plt.show()
```

### Insight
- The median screen time is almost similar for both males and females (around 4 hours).
- Both genders have high-value outliers above 8 hours.
- Overall, there is no major difference in average daily screen time between males and females.

### Screen time by Age bands

```python
#boxplot
sns.boxplot(x='age_group', y='avg_daily_screen_time_hr', data=df)
plt.xlabel("Age Bands")
plt.ylabel("Avg daily screen time(Hours)")
plt.title("Screen Time by Age Band")
plt.show()
```

### Insight
All age groups show similar average daily screen time around 4–5 hours, with 8–12 children showing slightly higher extreme values.

### Screen time by Location

```python
#barplot
sns.barplot(x='urban_or_rural', y='avg_daily_screen_time_hr', data=df)
plt.xlabel("Location")
plt.ylabel("Avg daily screen time(Hours)")
plt.title("Screen Time by Location Type")
plt.show()
```

### Insight
The average daily screen time is nearly the same for both urban and rural users, around 4–4.5 hours per day

### Screen time by age bands

```python
#Lineplot
sns.lineplot(x='age_group', y='avg_daily_screen_time_hr', data=df)
plt.xlabel("Age Band")
plt.ylabel("Average Screen Time (Hours)")
plt.title(" Screen Time Across Age Groups")
plt.show()
```

### Insight
The 8–12 children group has slightly lower average screen time, while 13–15 and 16–18 teens have slightly higher average screen time.
