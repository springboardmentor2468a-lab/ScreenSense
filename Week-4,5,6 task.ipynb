```python
def donut_chart(column_name, title):
    counts = df[column_name].value_counts()
    plt.figure()
    wedges, texts, autotexts = plt.pie(
        counts,
        labels=counts.index,
        autopct='%1.1f%%',
        pctdistance=0.75,     
        labeldistance=1.1      
    )
    centre_circle = plt.Circle((0, 0), 0.60, fc='white')
    fig = plt.gcf()
    fig.gca().add_artist(centre_circle)
    plt.title(title)
    plt.show()
```

```python
def classify_severity(text):
    if 'Obesity' in str(text):
        return 'High_Severity'
    elif 'Anxiety' in str(text):
        return 'Medium_Severity'
    elif 'Poor Sleep' in str(text) or 'Eye Strain' in str(text):
        return 'Low_Severity'
    else:
        return 'No_Impact'

df['Health_Severity'] = df['Health_Impacts'].apply(classify_severity)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
import seaborn as sns

sns.set_theme(style="whitegrid", font_scale=1.15)
plt.rcParams.update({"figure.dpi": 130, "savefig.bbox": "tight"})

df = pd.read_excel("C:\\Users\\meena\\Downloads\\screentime_clean.xlsx")
np.random.seed(42)
n = 800
df = pd.DataFrame({
    "Age":                               np.random.randint(8, 25, n),
    "Gender":                            np.random.choice(["Male","Female","Non-binary"], n, p=[0.47,0.47,0.06]),
    "Avg_Daily_Screen_Time_hr":          np.clip(np.random.normal(6.5, 2.5, n), 0.5, 16),
    "Primary_Device":                    np.random.choice(["Smartphone","Laptop","Tablet","Desktop","TV"], n, p=[0.40,0.25,0.15,0.10,0.10]),
    "Exceeded_Recommended_Limit":        np.random.choice(["Yes","No"], n, p=[0.45,0.55]),
    "Educational_to_Recreational_Ratio": np.clip(np.random.beta(2, 5, n), 0, 1),
    "Health_Impacts":                    np.random.choice(["None","Mild","Moderate","Severe"], n, p=[0.30,0.35,0.25,0.10]),
    "Urban_or_Rural":                    np.random.choice(["Urban","Rural"], n, p=[0.65,0.35]),
})

# ── Age Bands ──────────────────────────────────────────────────────
bins   = [7, 12, 15, 18, 24]
labels = ["Child (8-12)", "Early Teen (13-15)", "Late Teen (16-18)", "Young Adult (19-24)"]
df["Age_Band"] = pd.cut(df["Age"], bins=bins, labels=labels)

# ── Derived Feature 1: Device Mobility ────────────────────────────
df["Device_Mobility"] = df["Primary_Device"].apply(
    lambda d: "Portable" if d in ["Smartphone","Tablet","Laptop"] else "Wall-Mounted")

# ── Derived Feature 2: Usage Day Type (Weekend proxy) ─────────────
overall_mean = df["Avg_Daily_Screen_Time_hr"].mean()
df["Usage_Day_Type"] = df["Avg_Daily_Screen_Time_hr"].apply(
    lambda x: "High_Usage_Day (Weekend-like)" if x > overall_mean else "Normal_Usage_Day (Weekday-like)")

# ── Derived Feature 3: Education vs Recreational Category ─────────
df["Usage_Category"] = df["Educational_to_Recreational_Ratio"].apply(
    lambda r: "Education-Dominant" if r >= 0.5 else "Recreation-Dominant")

# ── Derived Feature 4: Device Preference Label ────────────────────
df["Device_Preference"] = df["Device_Mobility"].map({
    "Portable":     "Portable-Dominant",
    "Wall-Mounted": "Wall-Mounted-Dominant"})

# ── Derived Feature 5: Health Severity Score + Band ───────────────
df["Health_Severity_Score"] = df["Health_Impacts"].map(
    {"None": 0, "Mild": 1, "Moderate": 2, "Severe": 3})
df["Health_Severity_Band"] = df["Health_Severity_Score"].apply(
    lambda s: "Low (None)"      if s == 0 else
              "Mild (1)"        if s == 1 else
              "Moderate (2)"    if s == 2 else "Severe (3)")

# ── Shared constants ───────────────────────────────────────────────
AGE_ORDER      = ["Child (8-12)", "Early Teen (13-15)", "Late Teen (16-18)", "Young Adult (19-24)"]
HEALTH_ORDER   = ["None", "Mild", "Moderate", "Severe"]
HEALTH_COLORS  = ["#2ca02c", "#ffbf00", "#ff7f0e", "#d62728"]
SEVERITY_ORDER = ["Low (None)", "Mild (1)", "Moderate (2)", "Severe (3)"]

print("✅ Setup complete — shape:", df.shape)
df.head(3)
```
