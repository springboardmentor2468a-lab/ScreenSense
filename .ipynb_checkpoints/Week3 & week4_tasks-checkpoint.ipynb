### WEEK-2
### EDA - Exploratory Data Analysis
 ## Data Cleaning & Feature Engineering

```python
df["Gender"] = df["Gender"].fillna("Unknown")
df["Primary_Device"] = df["Primary_Device"].fillna("Unknown")
df["Health_Impacts"] = df["Health_Impacts"].fillna("None")
df["Urban_or_Rural"] = df["Urban_or_Rural"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
df["Avg_Daily_Screen_Time_hr"] = df["Avg_Daily_Screen_Time_hr"].fillna(df["Avg_Daily_Screen_Time_hr"].median())
```

## Why median?


#### Because screen time may have outlier

```python
df["Gender"] = df["Gender"].str.strip()
df["Primary_Device"] = df["Primary_Device"].str.strip()
df["Health_Impacts"] = df["Health_Impacts"].str.strip()
```

## Create Age Groups

```python
bins = [5,8,12,15,18]

labels = [
    "6-8",
    "9-12",
    "13-15",
    "16-18"
]

df["Age_Band"] = pd.cut(df["Age"], bins=bins, labels=labels)
```

## Create Screen Usage Category

```python
df["Screen_Time_Category"] = pd.cut(
    df["Avg_Daily_Screen_Time_hr"],
    bins=[0,2,4,10],
    labels=["Low","Moderate","High"]
)
```

### Low → <2 hours

### Moderate → 2–4 hours

### High → >4 hours

```python
### Educational vs Recreational Flag
df['Learning_Heavy_User'] = df['Educational_to_Recreational_Ratio'].apply(
    lambda x: 1 if x >= 0.5 else 0
)
```

```python
df['Day_Type'] = df['Avg_Daily_Screen_Time_hr'].apply(
    lambda x: 'Weekend' if x > 4 else 'Weekday'
)
```

```python
#Health Risk Indicator
df['Health_Risk_Flag'] = df['Avg_Daily_Screen_Time_hr'].apply(
    lambda x: 1 if x > 4 else 0
)
```

```python
df['User_Type'] = df['Educational_to_Recreational_Ratio'].apply(
    lambda x: 'Learning-Dominant' if x >= 0.5 else 'Recreation-Dominant'
)
```

# Save Cleaned Dataset

```python
df.to_csv("cleaned_kids_screen_time.csv", index=False)
```

### Insight example:

### Most children spend between 2 and 4 hours on screens daily.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="Avg_Daily_Screen_Time_hr",
    bins=15,
    kde=True,
    color="Darkblue"
)

plt.title("Distribution of Kids' Daily Screen Time", fontsize=14, fontweight='bold')
plt.xlabel("Average Daily Screen Time (Hours)")
plt.ylabel("Number of Children")

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

# Week 4: Device/Activity and Weekday/Weekend Analysis

```python
import squarify

device_counts = df['Primary_Device'].value_counts()

plt.figure(figsize=(8,6))

squarify.plot(
    sizes=device_counts.values,
    label=device_counts.index,
    alpha=0.8
)

plt.title("Overall Device Usage Distribution")

plt.axis('off')

plt.show()
```

**Smartphones represent the largest share of device usage among children.**

```python
plt.figure(figsize=(7,5))

sns.violinplot(
    x='Day_Type',
    y='Avg_Daily_Screen_Time_hr',
    data=df,
    palette="Set2"
)

plt.title("Weekday vs Weekend Screen Time Pattern")

plt.show()
```

**Screen time tends to increase during weekends compared to weekdays.**

```python
from statsmodels.graphics.mosaicplot import mosaic

plt.figure(figsize=(8,6))

mosaic(df, ['Gender','Primary_Device'])

plt.title("Device Mix Across Gender")

plt.show()
```

**Both genders predominantly use smartphones, though slight variations appear in tablet usage.**

```python
plt.figure(figsize=(8,5))

sns.kdeplot(
    data=df,
    x="Avg_Daily_Screen_Time_hr",
    hue="Urban_or_Rural",
    fill=True
)

plt.title("Screen Time Distribution by Location")

plt.show()
```

**Urban children show slightly higher screen-time density compared to rural children.**

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Age",
    y="Avg_Daily_Screen_Time_hr",
    size="Educational_to_Recreational_Ratio",
    hue="Primary_Device",
    sizes=(50,400),
    data=df
)

plt.title("Screen Time vs Age (Bubble = Learning Ratio)")

plt.show()

```

**Larger bubbles indicate children who spend a higher proportion of screen time on educational activities.**

```python
plt.figure(figsize=(9,5))

sns.swarmplot(
    x="Primary_Device",
    y="Avg_Daily_Screen_Time_hr",
    data=df
)

plt.xticks(rotation=30)

plt.title("Screen Time Distribution by Device")

plt.show()

```



**Smartphones display the widest spread of screen-time values, suggesting varied usage intensity.**

```python
g = sns.FacetGrid(df, row="Primary_Device", height=1.5, aspect=4)

g.map(sns.kdeplot, "Avg_Daily_Screen_Time_hr", fill=True)

g.set_titles("{row_name}")

plt.show()
```



**Screen time distributions vary across devices, with smartphones showing higher density at larger screen-time values.**

```python
plt.figure(figsize=(8,6))

plt.hexbin(
    df["Age"],
    df["Avg_Daily_Screen_Time_hr"],
    gridsize=20,
    cmap="Blues"
)

plt.colorbar(label="Number of Children")

plt.xlabel("Age")
plt.ylabel("Screen Time (Hours)")

plt.title("Peak Screen Time Cohorts")

plt.show()
```

**The densest clusters appear among children aged 13–15 with higher daily screen time.**

```python
sns.pairplot(
    df[['Age',
        'Avg_Daily_Screen_Time_hr',
        'Educational_to_Recreational_Ratio',
        'Gender']],
    hue='Gender',
    diag_kind='kde',
    palette='Set2'
)

plt.suptitle("Pairwise Relationship Between Screen Time Variables", y=1.02)

plt.show()
```

### Key Findings

1.Smartphones dominate device usage across most demographic groups.

2.Urban children exhibit slightly higher screen time compared to rural children.

3.Older children tend to spend more time on digital devices.

4.Recreation-dominant users display higher variability in screen-time behavior.

5.Weekend screen usage is higher than weekday patterns.

6.The peak screen-time cohort occurs among children aged 13–15 using smartphones.
