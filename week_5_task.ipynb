```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("cleaned_Indian_Kids_Screen_Time.csv")
```

```python
df.columns
```

```python
cohort = pd.crosstab(df['age_group'], df['primary_device'])
print(cohort)
```

### Age Band*Device

```python
cohort1 = pd.crosstab(df['age_group'], df['primary_device'])
sns.heatmap(cohort1, annot=True, fmt="d",cmap="coolwarm")
plt.title("Age Band vs Primary Device")
plt.xlabel("Primary Device")
plt.ylabel("Age Band")
plt.show()
```

### Gender*Activity Type

```python
cohort2 = pd.crosstab(df['gender'], df['activity_type'])
sns.heatmap(cohort2, annot=True, fmt="d",cmap='YlGnBu')
plt.title("Gender vs Activity Type")
plt.xlabel("Activity Type")
plt.ylabel("Gender")
plt.show()
```

### Location *Device

```python
cohort3 = pd.crosstab(df['urban_or_rural'], df['primary_device'])
sns.heatmap(cohort3, annot=True,fmt="d", cmap='viridis')
plt.title("Location vs Device Usage")
plt.xlabel("Primary Device")
plt.ylabel("Location Type")
plt.show()
```

### Age band * Weekday flag

```python
cohort4 = pd.crosstab(df['age_group'], df['weekday_flag'])
sns.heatmap(cohort4, annot=True,fmt="d", cmap='viridis')
plt.title("Age Band vs Weekday / Weekend")
plt.xlabel("Weekday Flag")
plt.ylabel("Age Band")
plt.show()
```

### Age Band * Device

```python
pd.crosstab(df['age_group'], df['primary_device']).plot(kind='bar', stacked=True,rot=360)
plt.title("Device Usage by Age Band")
plt.xlabel("Age Band")
plt.ylabel("Number of Users")
plt.show()
```

### Gender * Activity Type

```python
pd.crosstab(df['gender'], df['activity_type']).plot(kind='bar', stacked=True,rot=360)
plt.title("Activity type by  Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Users")
plt.show()
```

### Location * Activity

```python
pd.crosstab(df['urban_or_rural'], df['activity_type']).plot(kind='bar', stacked=True,rot=360)
plt.title("Activity Type by Location")
plt.xlabel("Location Type")
plt.ylabel("Number of Users")
plt.show()
```

### Device * Activity

```python
pd.crosstab(df['primary_device'], df['activity_type']).plot(kind='bar', stacked=True,rot=360)
plt.title("Activity Distribution by Device")
plt.xlabel("Primary Device")
plt.ylabel("Number of Users")
plt.show()
```
