```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("Indian_Kids_Screen_Time.csv")
```

```python
#univariate
#age-histogram
plt.figure()
plt.hist(data["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#average daily screentime-histogram 
plt.figure()
plt.hist(data["Avg_Daily_Screen_Time_hr"], bins=15)
plt.title("Screen Time Distribution")
plt.xlabel("Hours")
plt.ylabel("Frequency")
plt.show()
```

```python
#educational to recreatinal ratio-histogram
plt.figure()
plt.hist(data["Educational_to_Recreational_Ratio"], bins=10)
plt.title("Educational to Recreational Ratio Distribution")
plt.xlabel("Ratio")
plt.ylabel("Frequency")
plt.show()
```

```python
#gender-bar chart
plt.figure()
data["Gender"].value_counts().plot(kind="bar")
plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
#primary device-barchart
plt.figure()
data["Primary_Device"].value_counts().plot(kind="bar")
plt.title("Primary Device Usage")
plt.xlabel("Device")
plt.ylabel("Count")
plt.show()
```

```python
#exceeded recommended limit-bar chart
plt.figure()
data["Exceeded_Recommended_Limit"].value_counts().plot(kind="bar")
plt.title("Exceeded Recommended Limit")
plt.xlabel("Exceeded")
plt.ylabel("Count")
plt.show()
```

```python
#urban vs rural-barchart
plt.figure()
data["Urban_or_Rural"].value_counts().plot(kind="bar")
plt.title("Urban vs Rural Distribution")
plt.xlabel("Location")
plt.ylabel("Count")
plt.show()
```

```python
#bivariate analysis
#age vs screentime-scatter
plt.figure()
plt.scatter(data["Age"], data["Avg_Daily_Screen_Time_hr"])
plt.title("Age vs Screen Time")
plt.xlabel("Age")
plt.ylabel("Screen Time (Hours)")
plt.show()
```

```python
#Gender vs Screen Time (Boxplot)
plt.figure()
data.boxplot(column="Avg_Daily_Screen_Time_hr", by="Gender")
plt.title("Screen Time by Gender")
plt.suptitle("")
plt.xlabel("Gender")
plt.ylabel("Screen Time (Hours)")
plt.show()
```

```python
#Device vs Screen Time (Boxplot)
plt.figure()
data.boxplot(column="Avg_Daily_Screen_Time_hr", by="Primary_Device")
plt.title("Screen Time by Device")
plt.suptitle("")
plt.xlabel("Device")
plt.ylabel("Screen Time (Hours)")
plt.show()
```

```python
#Urban/Rural vs Screen Time (Boxplot)
plt.figure()
data.boxplot(column="Avg_Daily_Screen_Time_hr", by="Primary_Device")
plt.title("Screen Time by Device")
plt.suptitle("")
plt.xlabel("Device")
plt.ylabel("Screen Time (Hours)")
plt.show()
```

```python
#Educational Ratio vs Screen Time (Scatter)
plt.figure()
plt.scatter(data["Educational_to_Recreational_Ratio"],
            data["Avg_Daily_Screen_Time_hr"])
plt.title("Educational Ratio vs Screen Time")
plt.xlabel("Educational Ratio")
plt.ylabel("Screen Time (Hours)")
plt.show()
```

```python
#For Categorical Columns (Gender, Device, Location)
#Pie Chart
data["Primary_Device"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("Device Share")
plt.ylabel("")
plt.show()
```

```python
#Donut Chart
counts = data["Gender"].value_counts()

plt.pie(counts, labels=counts.index, autopct="%1.1f%%")
centre_circle = plt.Circle((0,0),0.70,fc='white')
fig = plt.gcf()
fig.gca().add_artist(centre_circle)
plt.title("Gender Distribution")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
data = pd.read_csv("Indian_Kids_Screen_Time.csv")
```

```python
#SINGLE NUMERICAL COLUMN
#Histogram (All Numerical Columns)
num_cols = ["Age",
            "Avg_Daily_Screen_Time_hr",
            "Educational_to_Recreational_Ratio"]
for col in num_cols:
    plt.figure()
    plt.hist(data[col], bins=10)
    plt.title(f"{col} Distribution")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
#Boxplot (All Numerical Columns)
for col in num_cols:
    plt.figure()
    plt.boxplot(data[col])
    plt.title(f"{col} Boxplot")
    plt.ylabel(col)
    plt.show()
```

```python
#Density Plot (KDE)
for col in num_cols:
    plt.figure()
    data[col].plot(kind="density")
    plt.title(f"{col} Density Plot")
    plt.show()
```

```python
#SINGLE CATEGORICAL COLUMN
#Bar Chart (All Categorical Columns)
cat_cols = ["Gender",
            "Primary_Device",
            "Exceeded_Recommended_Limit",
            "Urban_or_Rural"]

for col in cat_cols:
    plt.figure()
    data[col].value_counts().plot(kind="bar")
    plt.title(f"{col} Distribution")
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
#Pie Chart
for col in cat_cols:
    plt.figure()
    data[col].value_counts().plot(kind="pie", autopct="%1.1f%%")
    plt.title(f"{col} Share")
    plt.ylabel("")
    plt.show()
```

```python
#NUMERICAL vs NUMERICAL
#Scatter Plot (All Combinations)
plt.figure()
plt.scatter(data["Age"], data["Avg_Daily_Screen_Time_hr"])
plt.title("Age vs Screen Time")
plt.xlabel("Age")
plt.ylabel("Screen Time")
plt.show()
```

```python
#NUMERICAL vs CATEGORICAL
for col in cat_cols:
    plt.figure()
    sns.boxplot(x=data[col],
                y=data["Avg_Daily_Screen_Time_hr"])
    plt.title(f"Screen Time vs {col}")
    plt.show()
```

```python
#Bar Plot (Mean Comparison)
for col in cat_cols:
    plt.figure()
    data.groupby(col)["Avg_Daily_Screen_Time_hr"].mean().plot(kind="bar")
    plt.title(f"Average Screen Time by {col}")
    plt.ylabel("Average Hours")
    plt.show()
```

```python
#CATEGORICAL vs CATEGORICAL
#Grouped Bar Chart
pd.crosstab(data["Gender"],
            data["Primary_Device"]).plot(kind="bar")
plt.title("Gender vs Device")
plt.show()
```

```python
#stacked bar chart
pd.crosstab(data["Gender"],
            data["Primary_Device"]).plot(kind="bar", stacked=True)
plt.title("Gender vs Device (Stacked)")
plt.show()
```
