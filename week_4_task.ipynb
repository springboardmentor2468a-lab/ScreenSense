```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv("cleaned_Indian_Kids_Screen_Time.csv")
```

```python
df.columns
```

### Device usage by gender

```python
#Bar chart
sns.countplot(x='primary_device', hue='gender', data=df)
plt.xlabel("divice type")
plt.ylabel("Number of Users")
plt.title("Device Usage by Gender")
plt.show()
```

### Observation:
Smartphones are the most commonly used devices for both males and females.

### Device Usage by Age Band

```python
plt.figure(figsize=(8,5))
sns.countplot(x='age_group', hue='primary_device', data=df)
plt.title("Device Usage by Age Band")
plt.xlabel("Age Band")
plt.ylabel("Number of Users")
plt.show()
```

### Observation:
Smartphones dominate device usage in all age groups, while tablets are the least used devices.

### Devive Usage by Loaction

```python
sns.countplot(x='urban_or_rural', hue='primary_device', data=df)
plt.xlabel("location type")
plt.ylabel("no of users")
plt.title("Device Usage by Urban vs Rural")
plt.show()
```

### Observation
Urban users show higher device usage overall, while smartphones remain the most used device in both urban and rural areas.

# Activity Categories

```python
sns.countplot(x='activity_type', data=df)
plt.title("Distribution of Activity Types")
plt.xlabel("Activity Type")
plt.ylabel("Number of Users")
plt.xticks(rotation=360)
plt.show()
```

### Obsevarionv
Recreational usage is the primary activity type observed in the dataset, with a high number of users involved

### Acrivity category by gender

```python
plt.figure(figsize=(8,5))
sns.countplot(x='activity_type', hue='gender', data=df)
plt.title("Activity Category by Gender")
plt.xlabel("Activity Category")
plt.ylabel("Number of Users")
plt.xticks(rotation=360)
plt.show()
```

### Observation
Male users show slightly higher participation in recreational activities compared to female users, but the difference is minimal.

### Acrivity category by age band

```python
plt.figure(figsize=(8,5))
sns.countplot(x='age_group', hue='activity_type', data=df)
plt.title("Activity Category by Age Band")
plt.xlabel("Age group")
plt.ylabel("Number of Users")
plt.show()
```

### Observation
Recreational activity usage is highest among children (8–12) and older teens (16–18), while the 13–15 age group shows comparatively lower participation.

### Acrivity category by loaction

```python
plt.figure(figsize=(8,5))
sns.countplot(x='urban_or_rural', hue='activity_type', data=df)
plt.title("Activity Category by Day Type")
plt.xlabel("Activity Category")
plt.ylabel("Number of Users")
plt.xticks(rotation=360)
plt.show()
```

### Observation
Recreational activity usage is  higher in urban areas compared to rural areas.


# Week day vs Week end Analysis


### screen time on weekday vd weekends

```python
plt.figure(figsize=(8,5))
sns.boxplot(x='weekday_flag', y='avg_daily_screen_time_hr', data=df)
plt.title("Screen Time: Weekday vs Weekend")
plt.xlabel("Day Type")
plt.ylabel("Average Screen Time (Hours)")
plt.show()
```

### Observation
Users spend more time on screens during weekends compared to weekdays

### Activity CAtegory on weekday vs weekends

```python
plt.figure(figsize=(8,5))
sns.countplot(x='activity_type', hue='weekday_flag', data=df)
plt.title("Activity Category: Weekday vs Weekend")
plt.xlabel("Activity Category")
plt.ylabel("Number of Users")
plt.xticks(rotation=360)
plt.show()
```

### Observation
Recreational activity participation is slightly higher on weekends compared to weekdays.

### Device Usage on weekends vs weekends

```python
plt.figure(figsize=(8,5))
sns.countplot(x='primary_device', hue='weekday_flag', data=df)
plt.title("Device Usage: Weekday vs Weekend")
plt.xlabel("Device Type")
plt.ylabel("Number of Users")
plt.show()
```

### Observation
Smartphones are the most commonly used devices, and overall device usage is slightly higher on weekends than weekdays
