```python
AGE_ORDER = df['Age_Group'].cat.categories.tolist()

# Create a new column for usage category based on ratio
df['Usage_Category_Type'] = df['Educational_to_Recreational_Ratio'].apply(
    lambda x: 'Education-Dominant' if x > 1 else 'Recreation-Dominant'
)

# Which age group uses screens more educationally vs recreationally?
edu_age = df.groupby(["Age_Group", "Usage_Category_Type"], observed=True).size().unstack(fill_value=0)
edu_age = edu_age.reindex(AGE_ORDER)

fig, ax = plt.subplots(figsize=(10, 6))
edu_age.plot(kind="bar", ax=ax, color=["#9467bd", "#ff7f0e"],
             edgecolor="white", width=0.6)
ax.set_title("Education-Dominant vs Recreation-Dominant Usage by Age Group",
             fontsize=14, fontweight="bold")
ax.set_xlabel("Age Group")
ax.set_ylabel("Number of Users")
plt.xticks(rotation=45)
plt.legend(title="Usage Type")
plt.tight_layout()
plt.show()
```

```python
# Do education-dominant users screen more or less than recreation-dominant?
edu_screen = (df.groupby(["Age_Group", "Usage_Category_Type"], observed=True)
                ["Avg_Daily_Screen_Time_hr"].mean().unstack(fill_value=0))
edu_screen = edu_screen.reindex(AGE_ORDER)

fig, ax = plt.subplots(figsize=(10, 6))
edu_screen.plot(kind="bar", ax=ax, color=["#9467bd", "#ff7f0e"],
                edgecolor="white", width=0.6)
ax.set_title("Mean Screen Time: Education vs Recreation Users by Age Band",
             fontsize=14, fontweight="bold")
plt.show()
```

```python
# Do males/females differ in their device preference dominance?

# Create a new column for device preference
df['Device_Preference'] = df['Primary_Device'].apply(
    lambda x: 'Portable-Dominant' if x in ['Smartphone', 'Laptop'] else (
        'Wall-Mounted-Dominant' if x == 'TV' else 'Other'
    )
)

pref_gender = df.groupby(["Gender", "Device_Preference"]).size().unstack(fill_value=0)

fig, ax = plt.subplots(figsize=(9, 6))
pref_gender.plot(kind="bar", ax=ax, color=["#4878CF", "#6ACC65"],
                 edgecolor="white", width=0.5)
ax.set_title("Device Preference by Gender\n(Portable-Dominant vs Wall-Mounted-Dominant)",
             fontsize=14, fontweight="bold")
ax.set_xlabel("Gender")
ax.set_ylabel("Number of Users")
plt.show()
```

```python
# Which age group suffers most from screen time health impacts?

HEALTH_ORDER = df['Health_Impacts'].dropna().unique().tolist()
HEALTH_COLORS = sns.color_palette('viridis', n_colors=len(HEALTH_ORDER))

health_age = (df.groupby(["Age_Group", "Health_Impacts"], observed=True)
                .size().unstack(fill_value=0)[HEALTH_ORDER])
health_age = health_age.reindex(AGE_ORDER)

fig, ax = plt.subplots(figsize=(10, 6))
health_age.plot(kind="bar", ax=ax, color=HEALTH_COLORS,
                edgecolor="white", width=0.6)
ax.set_title("Health Impact Severity Distribution by Age Group",
             fontsize=14, fontweight="bold")
plt.show()
```

```python
# Do Wall-Mounted users have worse health than Portable users?
health_mob = (df.groupby(["Age_Band", "Device_Mobility"], observed=True)
                ["Health_Severity_Score"].mean().unstack(fill_value=0))
health_mob = health_mob.reindex(AGE_ORDER)

fig, ax = plt.subplots(figsize=(10, 6))
health_mob.plot(kind="bar", ax=ax, color=["#4878CF", "#d62728"],
                edgecolor="white", width=0.6)
ax.set_title("Mean Health Severity Score: Portable vs Wall-Mounted by Age Band\n(0=None · 1=Mild · 2=Moderate · 3=Severe)",
             fontsize=14, fontweight="bold")
plt.show()
```

```python
import numpy as np

def calculate_health_severity(impacts):
    if pd.isna(impacts) or impacts == 'None':
        return 0
    return len(impacts.split(', '))

df['Health_Severity_Score'] = df['Health_Impacts'].apply(calculate_health_severity)
```

```python
# Do Wall-Mounted users have worse health than Portable users?
health_mob = (df.groupby(["Age_Group", "Device_Preference"], observed=True)
                ["Health_Severity_Score"].mean().unstack(fill_value=0))
health_mob = health_mob.reindex(AGE_ORDER)

fig, ax = plt.subplots(figsize=(10, 6))
health_mob.plot(kind="bar", ax=ax, color=["#4878CF", "#d62728"],
                edgecolor="white", width=0.6)
ax.set_title("Mean Health Severity Score: Portable vs Wall-Mounted by Age Group\n(0=None · 1=Mild · 2=Moderate · 3=Severe)",
             fontsize=14, fontweight="bold")
ax.set_xlabel("Age Group")
ax.set_ylabel("Mean Health Severity Score")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# One summary chart showing mean screen time across all major segments
overall_mean = df["Avg_Daily_Screen_Time_hr"].mean()

segments = {
    "Age Group 6-8":   df[df["Age_Group"]=="6-8"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Age Group 9-11":  df[df["Age_Group"]=="9-11"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Age Group 12-14": df[df["Age_Group"]=="12-14"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Age Group 15-17": df[df["Age_Group"]=="15-17"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Male":                df[df["Gender"]=="Male"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Female":              df[df["Gender"]=="Female"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Urban":               df[df["Urban_or_Rural"]=="Urban"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Rural":               df[df["Urban_or_Rural"]=="Rural"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Portable-Dominant":   df[df["Device_Preference"]=="Portable-Dominant"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Wall-Mounted-Dominant": df[df["Device_Preference"]=="Wall-Mounted-Dominant"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Other Device Preference": df[df["Device_Preference"]=="Other"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Education-Dominant":  df[df["Usage_Category_Type"]=="Education-Dominant"]["Avg_Daily_Screen_Time_hr"].mean(),
    "Recreation-Dominant": df[df["Usage_Category_Type"]=="Recreation-Dominant"]["Avg_Daily_Screen_Time_hr"].mean()
}

seg_df = pd.DataFrame(list(segments.items()), columns=["Segment","Mean_Screen_Time"])
colors = sns.color_palette("tab10", n_colors=len(segments))

fig, ax = plt.subplots(figsize=(14, 7))
bars = ax.barh(seg_df["Segment"][::-1], seg_df["Mean_Screen_Time"][::-1],
               color=colors[::-1], edgecolor="white")
ax.bar_label(bars, fmt="%.2f hrs", padding=4, fontsize=10)
ax.axvline(overall_mean, color="navy",
           linestyle="--", linewidth=1.5, label=f"Overall Mean = {overall_mean:.2f} hrs")
ax.set_title("Segment-Wise Mean Screen Time Summary\n(Age · Gender · Location · Device · Usage Type)",
             fontsize=14, fontweight="bold")
ax.set_xlabel("Mean Screen Time (hrs)")
ax.legend()
ax.set_xlim(0, 10)
plt.tight_layout()
plt.show()
```
